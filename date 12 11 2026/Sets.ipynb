Set1={1,2,3,4,5}
Set1.add(el) -> add an element el
Set1.remove(el) -> Remove the element el
Set1.clear() -> Empties the Set
Set1.pop() -> Remove a random value

```python
Set1={1,2,3,4,5}
Set2={10,20,30}
Set1.add(60)
print(Set1) 
Set1.remove(1)
print(Set1) 
#Set1.clear()

Set1.pop()
print(Set1) 
Set1.clear()
print(Set1)
Set1.update(Set2)
print(Set1)
```
