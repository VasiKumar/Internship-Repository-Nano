Union - Two sets will get a the combination of all values with sets properties of course.

```python
set1={1,2,3,4,5}
set2={5,6,7}
x = set1.union(set2)
print(x)
```

Intersection- Two sets will get the intersection or the common values only any other values are removed.

```python
set1={1,2,3,4,5}
set2={5,6,7}
x = set1.intersection(set2)
print(x)
```

```python
a = {1, 2, 3, 4, 5}
b = {3, 4, 5, 6, 7}

result = a.difference(b)

# Compares what is unique to a which is not in b thats it.
print(result)
```

```python
set1={1,2,3,4,5,6}
set1.remove(1)
set1.discard(3)
print(set1)
```

Symmetric difference- only keeps not common ones.

```python
set1={3,4,5}
set2={4,5,6}
z=set1.symmetric_difference(set2)
print(z)
```

isdisjoint - Is the method to check if there exist any common or not. If common exist then output gives false. If no common exist then gives True

```python
set1={3,4,5}
set2={4,5,6}
x=set1.isdisjoint(set2)
print(x)
```

```python
set1={4,5}
set2={7,6}
x=set1.isdisjoint(set2)
print(x)
```
