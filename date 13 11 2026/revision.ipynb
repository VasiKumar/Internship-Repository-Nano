Union - Two sets will get a the combination of all values with sets properties of course.

```python
set1={1,2,3,4,5}
set2={5,6,7}
x = set1.union(set2)
print(x)
```

Intersection- Two sets will get the intersection or the common values only any other values are removed.

```python
set1={1,2,3,4,5}
set2={5,6,7}
x = set1.intersection(set2)
print(x)
```
