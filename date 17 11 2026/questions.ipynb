WAP to find the greatest number between 3 numbers taken as user input.

```python
n=int(input("Enter number 1: "))
m=int(input("Enter number 2: "))
p=int(input("Enter number 3: "))

if(n>m and n>p):
    print(f"{n} is greatest i.e. first number.")
if(m>n and m>p):
    print(f"{m} is greatest i.e. second number.")
else:
    print(f"{p} is greatest i.e. third number.")
```

Normally we have a switch case but here in Python we have match case pair. We use match which is just switch. Then case are similar. 
Then afterwards here there is no break needed here we just use case then its result and next case.

Note- There is underscore here in place if default. It actually is default which is underscore.

```python
day =5
match(day):
    case 1:
        print("Sunday")
        
    case 2:
        print("Monday")

    case 3:
        print("Tuesday")

    case 4:
        print("Wednesday")

    case 5:
        print("Thursday")

    case 6:
        print("Friday")

    case 7:
        print("Saturday")

    case _:
        print("Invalid day")
```
