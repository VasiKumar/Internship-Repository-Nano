WAP to find the number is even or odd.

```python
n= int(input("Enter a number: "))

if (n%2 ==0):
    print("Even")
else:
    print("Odd")
```

WAP to find the greatest number between 3 numbers taken as user input.

```python
n=int(input("Enter number 1: "))
m=int(input("Enter number 2: "))
p=int(input("Enter number 3: "))

if(n>m and n>p):
    print(f"{n} is greatest i.e. first number.")
if(m>n and m>p):
    print(f"{m} is greatest i.e. second number.")
else:
    print(f"{p} is greatest i.e. third number.")
```
