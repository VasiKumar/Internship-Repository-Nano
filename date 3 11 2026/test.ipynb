Write a program to find duplicate elements in a list and display their count.

```python
arr=[1,2,2,3,4,5]
flag=0
for i in range(len(arr)-2):
    if(arr[i]==arr[i+1]):
        flag=flag+1
print(flag)
```

Write a program to find the most frequently occurring word in a sentence.

```python
ab = "Basistha Basistha Basistha Basistha Basistha KK K K jd dh fajkas"

words = ab.split()

max_count = 0
most_word = ""

for w in words:
    count = words.count(w)
    if count > max_count:
        max_count = count
        most_word = w

print("Word:", most_word)
print("Count:", max_count)
```
